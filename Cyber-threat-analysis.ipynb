## **MEANING OF EVERY COLUMN**
id :
Unique identifier for each record (one row = one case/communication instance).

text : 
Raw textual content carried over the network (e.g., email/message body or payload). It may contain mentions of threats, actors, or attack patterns.

Entries (list of JSON objects) : 
Detected entities (and their spans) within text. Each object includes:

sender_id: ID of the entity that initiated/sent the communication.

label: Entity/threat type (e.g., malware, attack pattern, identity, benign, software attack, threat actor).

start_offset: Start character index of the entity mention in text.

end_offset: End character index of the entity mention in text.

receiver_ids: List of IDs for entities that received or were targeted.

relations (list of tuples)
Pairs of entity IDs describing directed relationships (source ID → target ID) between entities mentioned in Entries (e.g., “threat actor uses malware”, “malware targets identity”).

diagnosis
Human- or system-generated description/assessment of the identified cyber threat in this instance (what it is, impact, context).

solution
Recommended mitigations or actions to address the diagnosed threat (e.g., patches, controls, configurations, updates).

### **PRETRAITEMENT**

Suppression des colonnes inutiles, cleaning the text

```python
cols_to_keep = [
    "id",
    "text",
    "diagnosis",
    "solutions",
    "label_1",
    "label_2",
    "label_3",
]

df = df[cols_to_keep]
df.head(10)
```

```python
# 3. Définir les colonnes textuelles à nettoyer
text_cols = ["text", "diagnosis", "solutions"]

def clean_text(s):
    """Nettoyage simple d'un texte."""
    if pd.isna(s):
        return s
    # a) tout en minuscules
    s = s.lower()
    # b) enlever les retours à la ligne
    s = s.replace("\n", " ").replace("\r", " ")
    # c) supprimer les espaces multiples
    s = ' '.join(s.split())
    # d) enlever les espaces au début / à la fin
    s = s.strip()
    return s

for col in text_cols:
    df[col] = df[col].apply(clean_text)
```

```python
# 4. Supprimer les lignes inutilisables
# - lignes sans texte
# - (optionnel) lignes sans label_1 si tu choisis label_1 comme cible
df = df.dropna(subset=["text"])       # on veut au moins un texte
df = df.dropna(subset=["label_1"])    # cible non nulle pour entraîner un modèle
```

```python
# 5. Supprimer les doublons éventuels texte + label_1
df = df.drop_duplicates(subset=["text", "label_1"])

print("Shape après nettoyage :", df.shape)
```

```python
# 6. (Optionnel) sauvegarder le dataset pré-traité
df.to_csv("cyber_dataset_clean.csv", index=False)
```
